```python
# un ABB es como un arbol genealogico
# pero con numeros  
# 1. cada nodo tiene como maximo 2 hijos.
# 2. Izquierda: numeros menores.
# 3. Derecha: numeros mayores.


"""
             10
            /   \
            5     15
           / \   /  \
          3   7  12  20
"""
```

```python
#Leccion 1: creando nodo
# Un nodo es como una caja
# guarda: el valor (numero)
# guarda: dos flechas apuntando a sus hijos
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.izq = None  # Hijo izquierdo (números menores)
        self.der = None  # Hijo derecho (números mayores)

# Creamos un nodo simple
raiz = Nodo(10)
print(f"Nodo creado con valor: {raiz.valor}") 
print(f"Hijo izquierdo: {raiz.izq}")
print(f"Hijo derecho: {raiz.der}")
```

```python
# Desafio 1: crea un nodo con el valor 50
# y muestra su valor
raiz = Nodo(50)
print(f"Nodo creado con valor: {raiz.valor}") 
print(f"Nodo creado con valor: {raiz.izq}") 
print(f"Nodo creado con valor: {raiz.der}") 
```

```python
#Leccion 2: Conectando los nodos
#Vamos a crear un arbol pequeñito
"""
        10
       /  \
      5   15
"""
#Crear raiz:
raiz= Nodo(10)
#Crear hijos:
raiz.izq=Nodo(5)
raiz.der=Nodo(15)
print(f"Nodo creado con valor: {raiz.valor}") 
print(f"Hijo Izquierdo: {raiz.izq.valor}") 
print(f"Hijo Derecho: {raiz.der.valor}") 
```

```python
#Desafio 2:
#Crear un arbol con raiz 20
#Hijo izq 10, hijo der 30
#Imprime el resultado

raiz = Nodo(20)
raiz.izq=Nodo(10)
raiz.der= Nodo(30)
print(f"Nodo creado con valor: {raiz.valor}")
print(f"Hijo izquierdo: {raiz.izq.valor}")
print(f"Hijo derecho: {raiz.der.valor}")
```

```python
# Insertar valores
# Regla:
# Si el valor es menor ir a la izquierda
# Si el valor es mayor ir a la derecha
# Repetir hasta encontrar un espacio vacio
def insertar(raiz, valor):
    #si el arbol esta vacio, crea la raiz
    if raiz is None:
        return Nodo(valor)
    #si el valor es menor, ir a la izquierda
    if valor < raiz.valor:
        raiz.izq = insertar(raiz.izq, valor)
    #si el valor es mayor, ir a la derecha
    else:
        raiz.der=insertar(raiz.der, valor)
    
    return raiz

#Ejemplo construir este arbol:
 
""" 
       10
     /    \
    5      15
   /
  3
"""
raiz= None
raiz= insertar(raiz,10)
raiz = insertar(raiz,5)
raiz = insertar(raiz,15)
raiz = insertar(raiz,3)

print(f"Raiz: {raiz.valor}")
print(f"izquierda:{raiz.izq.valor}")
print(f"Derecha:{raiz.der.valor}")
print(f"izquierda de izquierda:{raiz.izq.izq.valor}")
```

```python
# Desafio 3:
# inserta los valor 25,8,30,6 en ese orden
# en arbol vacio en ese orden
raiz= None
raiz= insertar(raiz,25)
raiz = insertar(raiz,8)
raiz = insertar(raiz,30)
raiz = insertar(raiz,6)

print(f"Raiz: {raiz.valor}")
print(f"izquierda:{raiz.izq.valor}")
print(f"Derecha:{raiz.der.valor}")
print(f"izquierda de izquierda:{raiz.izq.izq.valor}")
```
